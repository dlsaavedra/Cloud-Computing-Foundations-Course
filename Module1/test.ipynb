```python
def increment(x):
    return x + 1

def test_increment():
    assert increment(3) == 4
    assert increment(5) == 6, "Increment failed"
    
    #Catch failures with try/except
    try:
        assert increment(-2) == -1 
    except AssertionError:
        print("Incrementing negative number failed")
        
```

```python
test_increment()
```

```python
def decrement(x):
    return x - 1

def test_decrement():
    assert decrement(3) == 2
    assert decrement(5) == 4, "Increment failed"
    
    #Catch failures with try/except
    try:
        assert decrement(-2) == -3 
    except AssertionError:
        print("Incrementing negative number failed")
```

```python
test_increment()
```

# Second Function

```python
def square(x):
    return x**2

def test_square():
    assert square(3) == 9
    assert square(5) == 25, "Increment failed"
    
    #Catch failures with try/except
    try:
        assert square(-2) == 4
    except AssertionError:
        print("Incrementing negative number failed")
```

```python
test_square()
```

```python
# 3. Crear una aserción personalizada para validar umbrales
def assert_dentro_de_umbral(valor, minimo, maximo):
    assert minimo <= valor <= maximo, (
        f"{valor} está fuera del intervalo [{minimo}, {maximo}]"
    )
    

def test_increment_dentro_de_umbral():
    resultado = increment(3)
    assert_dentro_de_umbral(resultado, minimo=3, maximo=5)

test_increment_dentro_de_umbral()
```

```python
import ipytest
import pytest

# 1. Reset the internal ipytest registry so it forgets tests from other cells
ipytest.clean()


# 4. Usar una función auxiliar para generar datos de prueba
def generar_pares_incremento(valores):
    return [(valor, valor + 1) for valor in valores]


@pytest.mark.parametrize(
    "entrada, esperado",
    generar_pares_incremento([-3, 0, 8]),
)
def test_increment_datos_generados(entrada, esperado):
    assert increment(entrada) == esperado

   
ipytest.run("-v")
```

```python
# 5. Verificar una excepción esperada
def test_increment_tipo_invalido():
    with pytest.raises(TypeError, match="x debe ser un entero"):
        increment("3")
        
test_increment_tipo_invalido()
```
